```python
import hyperspy.api as hs
import lumispy
from pathlib import Path
import matplotlib.pyplot as plt
```

```python


home_path = Path.home()

data_root = home_path / 'Library/CloudStorage/OneDrive-UniversityofCambridge/Desktop/OneDrive/project_data/raw/2026/04_02_2026/ZEKI-INP-SPAD'

for path in data_root.rglob("*.sur"):
    s = hs.load(path)
    s.plot()
    
    
    # Mirror the folder structure under ../plots/
    folder_name = path.parent.name
    save_dir = Path("../plots") / folder_name
    save_dir.mkdir(parents=True, exist_ok=True)
    
    plt.savefig(save_dir / f"{path.stem}.png")
    plt.close()
```
